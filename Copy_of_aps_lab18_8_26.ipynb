```python
#the birthday problem
#probability that at least 2 people have the same birthday
import random

#no. of simulations for each count
trials=10000

people_count=[]
probability=[]

#no. of people from 3 to 100
for people in range(3,101):
  shared_birthdays=0
  for trial in range(trials):
    birthdays=[]
    for ppl in range(people):
      birthday=random.randint(1,365)
      birthdays.append(birthday)
    #to find number of common birthdays
    if len(birthdays)!=len(set(birthdays)):
      shared_birthdays+=1
  people_count.append(people)
  probability.append((shared_birthdays/trials)*100)

  print("No. of people:",people,"Probability:",shared_birthdays/trials)


```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 5))
plt.plot(
    people_count,
    probability,
    color="blue",
    linewidth=2
)

plt.xlabel("No. of people")
plt.ylabel("Probability of at least 2 ppl sharing a birthday (%)")
plt.title("Birthday Problem")

#plt.grid(True, alpha=0.2)

plt.show()
```
