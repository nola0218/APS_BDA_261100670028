```python
#Problem 1
import random

#function to simulate flipping a coin
# 0 -> H, 1 -> T
def tossCoin():
  return random.randint(0, 1)

#flipping a coin 100 times and recording observations
cnt=[]
for i in range(100):
  flip = tossCoin()
  if flip == 0:
    cnt.append('H')
  else:
    cnt.append('T')
print("Number of heads: ", cnt.count("H"))
print("NUmber of tails: ", cnt.count("T"))
print("Experimental Probability of head: ", cnt.count("H")/100)
print("Experimental Probability of tail: ", cnt.count("T")/100)
```

```python
#Problem 2

#flipping coin n times and recording observations
def experiment(n):
  cnt=[]
  for i in range(n):
    flip = tossCoin()
    if flip == 0:
      cnt.append('H')
    else:
      cnt.append('T')
  print("For n =",n)
  print("Number of heads:", cnt.count("H"))
  print("NUmber of tails:", cnt.count("T"))
  print("Experimental Probability of head:", cnt.count("H")/n)
  print("Experimental Probability of tail:", cnt.count("T")/n)
  print()

#for n=10
n=10
experiment(n)

#for n=100
n=100
experiment(n)

#for n=1000
n=1000
experiment(n)

#for n=10000
n=10000
experiment(n)

#experimental probability almost equals to theoritical probability as n increases
print("experimental probability almost equals to theoritical probability as n increases")
```

```python
#Problem 3
import random

#function to simulate rolling a dice
def rollDice():
  return random.randint(1, 6)

#rolling a dice 6000 times
cnt=[]
for i in range(6000):
  roll = rollDice()
  cnt.append(roll)
print("Probability of getting 1:", cnt.count(1)/6000)
print("Probability of getting 2:", cnt.count(2)/6000)
print("Probability of getting 3:", cnt.count(3)/6000)
print("Probability of getting 4:", cnt.count(4)/6000)
print("Probability of getting 5:", cnt.count(5)/6000)
print("Probability of getting 6:", cnt.count(6)/6000)
```

```python
#Problem 4
cnt=[]
for i in range(10000):
  roll = rollDice()
  cnt.append(roll)
A=cnt.count(1)
B=cnt.count(6)
AUB=cnt.count(1)+cnt.count(6)
print("Probability of getting 1 (P(A)):", A/10000)
print("Probability of getting 6 (P(B)):", B/10000)
print("Probability of getting 1 or 6 (P(AUB)):", AUB/10000)
print("P(A) + P(B) =",(A/10000)+(B/10000))
#A and B are disjoint
#P(A)+P(B)=P(AUB), hence verified
print()
print("A and B are disjoint")
print("P(A)+P(B)=P(AUB), hence verified")
```

```python
#Problem 6
import random

#function to simulate tossing a coin
# 0 -> H, 1 -> T
def tossCoin():
  return random.randint(0, 1)

#function to simulate rolling a dice
def rollDice():
  return random.randint(1, 6)

res=[]
for i in range(10000):
  toss=tossCoin()
  roll=rollDice()
  res.append((toss, roll))
coinRes=[n[0] for n in res]
diceRes=[n[1] for n in res]
A=coinRes.count(0)
B=diceRes.count(2)+diceRes.count(4)+diceRes.count(6)
AnB=[n[0]==0 and (n[1]==2 or n[1]==4 or n[1]==6)  for n in res].count(True)
print("Probability of getting a head P(A):", A/10000)
print("Probability of getting an even number P(B):", B/10000)
print("Probability of getting a head and an even number P(AnB):", AnB/10000)
print("P(A)P(B) =", (A/10000)*(B/10000))
#P(AnB) = P(A)P(B), hence verified
print()
print("P(AnB) = P(A)P(B), hence verified")
```
